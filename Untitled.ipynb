```python
neighborhood_group = check_df.groupby('neighbourhood_cleansed')
```

```python
neighborhood_centers = neighborhood_group[['latitude', 'longitude']].median().reset_index()
```

```python
coords_rad = np.radians(neighborhood_centers[['latitude', 'longitude']])
```

```python
from sklearn.neighbors import NearestNeighbors
```

```python
nbrs = NearestNeighbors(n_neighbors=7, metric='haversine').fit(coords_rad)
distances, indices = nbrs.kneighbors(coords_rad)
```

```python
neighbor_map = {}
for i, row in neighborhood_centers.iterrows():
    neighborhood = row['neighbourhood_cleansed']
    # Extract the 6 nearest neighborhoods (excluding the neighborhood itself at index 0)
    neighbors = [neighborhood_centers.iloc[idx]['neighbourhood_cleansed'] for idx in indices[i][1:]]
    neighbor_map[neighborhood] = neighbors

# --- Step 5: Merge the 'nearest_neighbourhoods' column back into your 'check_df' ---
check_df['nearest_neighbourhoods'] = check_df['neighbourhood_cleansed'].map(neighbor_map)

# Check the results
print(check_df[['neighbourhood_cleansed', 'nearest_neighbourhoods']].head(10))
```
